## Подготовка данных

```python
# Создаём синтетические данные
np.random.seed(42)
dates = pd.date_range('2023-01-01', periods=365, freq='D')
y = 100 + np.cumsum(np.random.randn(365)) + 20 * np.sin(np.arange(365) / 7 * 2 * np.pi)

df = pd.DataFrame({
    'unique_id': 'series_1',
    'ds': dates,
    'y': y
})

print(df.head())
print(f"\nВсего точек: {len(df)}")
```

## Визуализация (шаблон)

```python
import matplotlib.pyplot as plt

def plot_timegpt_forecast(history_df, forecast_df, title=''):
    """Визуализация прогноза TimeGPT."""
    
    fig, ax = plt.subplots(figsize=(12, 5))
    
    # История
    ax.plot(history_df['ds'], history_df['y'], 'b-', label='История')
    
    # Прогноз
    ax.plot(forecast_df['ds'], forecast_df['TimeGPT'], 
            'r-', label='TimeGPT прогноз')
    
    # Интервалы (если есть)
    if 'TimeGPT-lo-90' in forecast_df.columns:
        ax.fill_between(
            forecast_df['ds'],
            forecast_df['TimeGPT-lo-90'],
            forecast_df['TimeGPT-hi-90'],
            alpha=0.3, color='red', label='90% интервал'
        )
    
    ax.axvline(x=history_df['ds'].iloc[-1], color='gray', linestyle='--', alpha=0.5)
    ax.legend()
    ax.set_title(title or 'TimeGPT: прогноз')
    ax.set_xlabel('Дата')
    ax.set_ylabel('Значение')
    ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()

# Демонстрация с синтетическими данными
# Имитируем прогноз
horizon = 24
forecast_dates = pd.date_range(df['ds'].iloc[-1] + pd.Timedelta(days=1), periods=horizon, freq='D')
forecast_values = df['y'].iloc[-1] + np.cumsum(np.random.randn(horizon) * 2)

fake_forecast = pd.DataFrame({
    'ds': forecast_dates,
    'TimeGPT': forecast_values,
    'TimeGPT-lo-90': forecast_values - 15,
    'TimeGPT-hi-90': forecast_values + 15,
})

plot_timegpt_forecast(df.tail(100), fake_forecast, 
                      title='TimeGPT: пример прогноза (синтетический)')
```
